```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df["posted_at"] = pd.to_datetime(df["posted_at"], dayfirst=True)
```

```python
df.isnull().sum()
```

```python
df["sentiment"] = df["sentiment"].fillna(df["sentiment"].mode()[0])
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
```

```python
df.isnull().sum()
```

```python
#checking for the outliers
plt.figure(figsize=(3,3))
sns.boxplot(x=df["age"])
plt.show()

plt.figure(figsize=(3,3))
sns.boxplot(x=df["likes"])
plt.show()

plt.figure(figsize=(3,3))
sns.boxplot(x=df["comments"])
plt.show()

plt.figure(figsize=(3,3))
sns.boxplot(x=df["shares"])
plt.show()
```

```python
# Filling missing values using mean
df["age"] = df["age"].fillna(df["age"].mean())
df["likes"] = df["likes"].fillna(df["likes"].mean())
df["comments"] = df["comments"].fillna(df["comments"].mean())
df["shares"] = df["shares"].fillna(df["shares"].mean())
```

```python
df.isnull().sum()
```

```python
#checking for duplicates
df.duplicated().sum()
```

```python
df.head()
```

```python
df["hashtag_count"] = df["hashtags"].str.split().str.len()
```

```python
#how sentimental contens influence people
sentiment_trend = df.groupby("sentiment")["impression_count"].mean()
plt.figure(figsize=(10,3))
plt.plot(sentiment_trend)

plt.title("influence by sentimental contents ")
plt.xlabel("sentiment category")
plt.ylabel("impression count Engagement rate")

plt.show()
```
